```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')
scaler = StandardScaler()

# 1. 데이터 Loading
data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Social_Network_Ads.csv')
age = data['Age'].values.reshape(-1,1)
esalary = data['EstimatedSalary'].values.reshape(-1,1)
pur = data['Purchased'].values.reshape(-1,1)
# 2. 데이터 전처리, Feature 추출
age = scaler.fit_transform(age)
esalary = scaler.fit_transform(esalary)
X = np.hstack([age**0,age, esalary])
y = pur
w = np.zeros([3,1])
X = np.asmatrix(X)
y = np.asmatrix(y)
w = np.asmatrix(w)
# 3. 학습 데이터 분할
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
C1 = np.where(y_train == 1)[0]
C0 = np.where(y_train == 0)[0]
C1_t = np.where(y_test == 1)[0]
C0_t = np.where(y_test == 0)[0]
# 4. 학습데이터에 대해 로지스틱회귀 모델 구성
def sigmoid(x, w):
  return 1 / (1 + np.exp(-x*w))
def cost_function(y, X, w):
  return -1 * sum(y*np.log(sigmoid(X,w)) + (1-y)*np.log(1-sigmoid(X,w)))
# 5. 경사하강법을 이용해서 모델 학습하고, 파라미터 값 출력
a = 0.01
epoch = 100000
for i in range(epoch):
  df = -X_train.T*(y_train - sigmoid(X_train,w))
  w = w - a*df
print(w[0,0], w[1,0], w[2,0])
# 6. 모델의 시각화
# 훈련 데이터 시각화
xp = np.linspace(np.min(age), np.max(age), 1000).reshape(-1,1)
yp = -1 * (w[1,0]/w[2,0]*xp + w[0,0]/w[2,0])
plt.figure(figsize = (6, 4))
plt.plot(X_train[C1,1], X_train[C1,2], 'go', markersize = 4, label = 'Class 1')
plt.plot(X_train[C0,1], X_train[C0,2], 'bo', markersize = 4, label = 'Class 0')
plt.plot(xp, yp, 'r', linewidth = 4, label = 'Dicision Boundary')
plt.title('Logistic Regression (Traning Data)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.ylim([np.min(esalary), np.max(esalary)])
plt.legend(loc = 1)
plt.show()
# 테스트 데이터 시각화
xp = np.linspace(np.min(age), np.max(age), 1000).reshape(-1,1)
yp = -1 * (w[1,0]/w[2,0]*xp + w[0,0]/w[2,0])
plt.figure(figsize = (6, 4))
plt.plot(X_test[C1_t,1], X_test[C1_t,2], 'go', markersize = 4, label = 'Class 1')
plt.plot(X_test[C0_t,1], X_test[C0_t,2], 'bo', markersize = 4, label = 'Class 0')
plt.plot(xp, yp, 'r', linewidth = 4, label = 'Dicision Boundary')
plt.ylim([np.min(esalary), np.max(esalary)])
plt.title('Logistic Regression (Traning Data)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend(loc = 1)
plt.show()
```
